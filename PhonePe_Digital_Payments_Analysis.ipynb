```python
#Task 3 Data Quality Checks 
# 3.1 Ensure data consistency across state and district levels.
 #3.1.1 for each state, calculate total number of transactions, total transaction amonunts and total registered users by summing up the values form
      # district  the district level data
#3.1.2 compare the result with corresponding values at state lavel to ensure they match
#3.1.3 display any discripancies found between district level and state level data
state_df = pd.read_excel(file_path, sheet_name="State_Txn and Users")
district_df = pd.read_excel(file_path, sheet_name="District_Txn and Users")

# 3.1.1 Aggregate district-level data
district_summary = (
    district_df
    .groupby(["State", "Year", "Quarter"])
    [["Transactions", "Amount (INR)", "Registered Users"]]
    .sum()
    .reset_index()
)
#print("District Level Aggregated Data")
#print(district_summary.head())

# 3.1.2 Compare with state-level data

state_summary = state_df[
    ["State", "Year", "Quarter",
     "Transactions", "Amount (INR)", "Registered Users"]
]

comparison = pd.merge(
    district_summary,
    state_summary,
    on=["State", "Year", "Quarter"],
    suffixes=("_District", "_State")
)
# 3.1.3 Find discrepancies

comparison["Txn_Diff"] = (
    comparison["Transactions_District"]
    - comparison["Transactions_State"]
)

comparison["Amount_Diff"] = (
    comparison["Amount (INR)_District"]
    - comparison["Amount (INR)_State"]
)

comparison["Users_Diff"] = (
    comparison["Registered Users_District"]
    - comparison["Registered Users_State"]
)

discrepancies = comparison[
    (comparison["Txn_Diff"] != 0) |
    (comparison["Amount_Diff"] != 0) |
    (comparison["Users_Diff"] != 0)
]

print("\nDiscrepancies Found:")
print(discrepancies)

print("\nTotal Discrepancies:", len(discrepancies))
```

```python
#4 Data Mergining and advance analysis
#4.1 Ratio of user to population by state
    #4.1.1. Merge the State_Txn and Users dataset with District Demographics dataset 
            #to calculate the ratio of registered users to the population for each state. display the result in tabular format
    #4.1.2. create a column chart depicting the ratio of users to population by state.

state_Txn = pd.read_excel(file_path, sheet_name="State_Txn and Users")
district_demo = pd.read_excel(file_path, sheet_name="District Demographics")


State_Regiter_User_Summary = (state_Txn
                     .groupby("State")["Registered Users"]
                     .sum()
                      .reset_index())


District_Demo_Summary = (
                        district_demo
                        .groupby("State")["Population"]
                        .sum()
                        .reset_index())

Merge_Data = pd.merge(
    State_Regiter_User_Summary,
    District_Demo_Summary,
    on ="State")

Merge_Data["User_to_Population_Ratio"]  = (Merge_Data["Registered Users"]/ Merge_Data["Population"])

print(Merge_Data.to_string(index = False))

plt.figure(figsize =(16,7))
sns.barplot(Merge_Data, x="State", y="User_to_Population_Ratio", color ="green")
plt.title("Ratio of Registered Users to Population by State")
plt.xlabel("State")
plt.ylabel("User to Population Ratio")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

```
